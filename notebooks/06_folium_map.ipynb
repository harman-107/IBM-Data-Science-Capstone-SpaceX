# 06 — Interactive Visual Analytics with Folium

IBM Data Science Capstone — SpaceX Falcon 9 first-stage landing prediction.

Create interactive maps of Falcon 9 launch sites and landing outcomes. Launch-site coordinates are already present in the API extract.

```python
import pandas as pd
import folium
from folium.plugins import MarkerCluster

df=pd.read_csv('../data/dataset_part_1.csv')
center=[28.57,-90]
m=folium.Map(location=center, zoom_start=3)
cluster=MarkerCluster().add_to(m)
for _,r in df.iterrows():
    folium.Marker([r.Latitude,r.Longitude], popup=f"Flight {r.FlightNumber}: {r.Outcome}").add_to(cluster)
m.save('../outputs/folium_launch_outcomes.html')
m
```

The capstone interpretation is that launch sites are coastal and relatively near the equator. The project also records proximity analysis for coastline, highway, railway and city features in the supplied map outputs.
