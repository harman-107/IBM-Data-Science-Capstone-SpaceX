# 03 — Data Wrangling

IBM Data Science Capstone — SpaceX Falcon 9 first-stage landing prediction.

Prepare the 90-row Falcon 9 dataset, create the binary landing target, impute missing payload mass, and preserve a modeling-ready copy.

```python
import pandas as pd
from pathlib import Path

ROOT=Path('..')
df=pd.read_csv(ROOT/'data/dataset_part_1.csv')
df.info()
```

```python
df['PayloadMass']=df['PayloadMass'].fillna(df['PayloadMass'].mean())
df['Class']=df['Outcome'].str.startswith('True').astype(int)
model_df=df.drop(columns=['Outcome','Date','BoosterVersion'])
model_df.to_csv(ROOT/'data/dataset_part_2.csv', index=False)
print('shape:', model_df.shape)
print('success rate:', model_df['Class'].mean())
print('missing values:', int(model_df.isna().sum().sum()))
```

The resulting target is `Class = 1` for a successful first-stage landing outcome and `0` otherwise. The study contains 90 Falcon 9 records and an overall success rate of 66.7%.
