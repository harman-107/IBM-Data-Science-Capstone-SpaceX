# 02 — Data Collection: Wikipedia Web Scraping

IBM Data Science Capstone — SpaceX Falcon 9 first-stage landing prediction.

Scrape the Wikipedia launch-history page with `pandas.read_html`, normalize the table headers, and retain Falcon 9 records through the capstone study period. The current Wikipedia page has been reorganized over time, so the code is deliberately defensive and the historical 90-row API extract remains the modeling source.

```python
import pandas as pd
URL='https://en.wikipedia.org/wiki/List_of_Falcon_9_and_Falcon_Heavy_launches'
tables=pd.read_html(URL)
len(tables), [t.shape for t in tables[:5]]
```

```python
# Inspect candidate tables and select rows containing Falcon 9 / F9.
candidates=[]
for i,t in enumerate(tables):
    s=t.astype(str).agg(' | '.join, axis=1)
    if s.str.contains('Falcon 9|F9', case=False, regex=True).any():
        candidates.append((i,t.shape))
candidates[:20]
```
