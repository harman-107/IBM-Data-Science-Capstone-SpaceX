# 07 — Plotly Dash Dashboard Analysis

IBM Data Science Capstone — SpaceX Falcon 9 first-stage landing prediction.

Prepare the dashboard dataset and the three stakeholder views required by the capstone: success/failure by launch site, payload range by site, and payload vs landing outcome.

```python
import pandas as pd
df=pd.read_csv('../data/dataset_part_1.csv')
df['class']=df['Outcome'].str.startswith('True').astype(int)
dash=df[['FlightNumber','LaunchSite','class','PayloadMass','BoosterVersion']].copy()
dash['Booster Version Category']=dash['BoosterVersion'].str.extract(r'(B5|FT|v1.1|v1.0)',expand=False).fillna('Other')
dash.to_csv('../data/spacex_launch_dash.csv',index=False)
dash.head()
```

Run `dashboard/app.py` to launch the interactive Dash application locally. The dashboard implements a launch-site selector and payload-mass range filter.
