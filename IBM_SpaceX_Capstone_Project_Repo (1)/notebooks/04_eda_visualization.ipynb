# 04 — EDA: Visualization

IBM Data Science Capstone — SpaceX Falcon 9 first-stage landing prediction.

```python
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv('../data/dataset_part_1.csv')
df['Class']=df['Outcome'].str.startswith('True').astype(int)
```

The following analyses reproduce the capstone visual questions: flight number vs launch site, payload vs launch site, success rate vs orbit, flight number vs orbit, payload vs orbit, and yearly launch-success trend.

```python
site=df.groupby('LaunchSite')['Class'].agg(['count','sum'])
print(site.assign(success_rate=site['sum']/site['count']))
```

```python
orbit=df.groupby('Orbit')['Class'].mean().sort_values(ascending=False)
print((orbit*100).round(2))
```

```python
df['Year']=pd.to_datetime(df['Date']).dt.year
print(df.groupby('Year')['Class'].mean().mul(100).round(1))
```
