IBM Data Science Capstone — SpaceX Falcon 9 first-stage landing prediction.

Run the required SQL-style exploratory queries with SQLite. The course SQL lab uses a related launch-outcome table; the repository keeps the standard course answer set in `docs/sql_reference_results.md` and demonstrates equivalent SQL patterns below.

```python
import sqlite3, pandas as pd
df=pd.read_csv('../data/dataset_part_1.csv')
df['Class']=df['Outcome'].str.startswith('True').astype(int)
con=sqlite3.connect(':memory:')
df.to_sql('SPACEXTBL',con,index=False,if_exists='replace')
```

```python
queries={
'launch_sites': 'SELECT DISTINCT LaunchSite FROM SPACEXTBL;',
'cca_rows': "SELECT * FROM SPACEXTBL WHERE LaunchSite LIKE 'CCAFS%' LIMIT 5;",
'payload_total': 'SELECT SUM(PayloadMass) AS total_payload_kg FROM SPACEXTBL;',
'avg_f9v1_1': "SELECT AVG(PayloadMass) AS avg_payload_kg FROM SPACEXTBL WHERE BoosterVersion='Falcon 9';"
}
for name,q in queries.items():
    print(name)
    display(pd.read_sql_query(q,con))
```
